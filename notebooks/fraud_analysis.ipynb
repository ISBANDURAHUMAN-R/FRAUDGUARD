## 1. Executive Summary & Objective
The goal of this analysis is to evaluate an extreme class imbalance classification problem (0.17% fraud rate), develop robust data pipelines without data leakage, compare Logistic Regression and Random Forest models, and optimize detection thresholds balancing high recall with practical precision.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

plt.style.use('dark_background')
print('Libraries successfully imported!')
```

## 2. Dataset Loading & Schema Validation

```python
data_path = Path('../data/creditcard.csv')
df = pd.read_csv(data_path)
print(f'Dataset Shape: {df.shape}')
print('Missing Values:', df.isnull().sum().max())
df.head()
```

## 3. Class Imbalance Examination
Target distribution: 284,315 legitimate transactions (99.83%) and 492 fraud cases (0.17%).

```python
class_counts = df['Class'].value_counts()
print(class_counts)
print(f'Fraud Percentage: {class_counts[1] / len(df) * 100:.4f}%')

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(x='Class', data=df, palette=['#00e676', '#ff5252'], ax=ax)
ax.set_yscale('log')
ax.set_title('Class Distribution Log Scale')
plt.show()
```

## 4. Amount & Time Distribution Analysis
Comparison of monetary transaction amounts and temporal occurrence between legitimate and fraudulent transactions.

```python
print('Legitimate Amount Stats:')
print(df[df['Class'] == 0]['Amount'].describe())
print('\nFraudulent Amount Stats:')
print(df[df['Class'] == 1]['Amount'].describe())

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(df[df['Class'] == 0]['Amount'], bins=50, ax=ax1, color='#00e676', log_scale=True)
ax1.set_title('Legitimate Amounts (Log Scale)')
sns.histplot(df[df['Class'] == 1]['Amount'], bins=50, ax=ax2, color='#ff5252', log_scale=True)
ax2.set_title('Fraudulent Amounts (Log Scale)')
plt.show()
```
